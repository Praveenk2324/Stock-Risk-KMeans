```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv(r"..\data\raw\Nifty_50_1y_dataset.csv", header=[0,1], index_col=0)
df.head(10)
```

```python
close_prices = df.xs(key='Close', axis=1, level=1).copy()
close_prices
```

```python
close_prices.isnull().sum()
```

```python
close_prices.drop(columns=['LTIM.NS', 'TATAMOTORS.NS'], inplace=True)
```

```python
close_prices.isnull().sum()
```

```python
close_prices = close_prices.ffill().dropna(axis=0)
```

```python
daily_returns = close_prices.pct_change().dropna()
```

```python
annual_returns = daily_returns.mean() * 252
```

```python
annual_volatility = daily_returns.std() * np.sqrt(252)
```

```python
clustering_data = pd.DataFrame({
    'Return': annual_returns,
    'Volatility': annual_volatility
})
```

```python
scaler = StandardScaler()
scaled_features = scaler.fit_transform(clustering_data)
```

```python
k_means_ready = pd.DataFrame(
    scaled_features,
    index=clustering_data.index,
    columns=clustering_data.columns
)
```
